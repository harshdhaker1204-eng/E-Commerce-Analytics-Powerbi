```python
#we are fill this data unknow becease this data have a categorical data.
dataset['Item']=dataset['Item'].fillna('Unknow')
```

```python
dataset["Price Per Unit"].median()
```

```python
#in this column we fill median value because this data is a numerical and we are going to fill median value
dataset["Price Per Unit"]=dataset["Price Per Unit"].fillna(dataset["Price Per Unit"].median())
```

```python
dataset["Quantity"].median()
```

```python
dataset["Quantity"]=dataset["Quantity"].fillna(dataset["Quantity"].median())
```

```python
dataset["Total Spent"].median()
```

```python
dataset["Total Spent"]=dataset["Total Spent"].fillna(dataset["Price Per Unit"] * dataset["Quantity"])
```

```python
dataset["Discount Applied"].mode()
```

```python
dataset['Discount Applied']=dataset['Discount Applied'].fillna(dataset['Discount Applied'].mode()[0])
```

```python
dataset.isnull().sum()
```

```python
# convert Transaction date currently it is object,but it should be a date

dataset["Transaction Date"]=pd.to_datetime(dataset["Transaction Date"],errors="coerce")
```

```python
dataset["Transaction Date"].dtype
```

```python
#check whether it contain decimal values
dataset["Quantity"].unique()
```

```python
dataset["Quantity"]=dataset["Quantity"].astype("Int64")
```

```python
dataset["Quantity"]
```

```python
#Now i will convert table name into snake name 
dataset.columns=(dataset.columns
                .str.strip()
                .str.lower()
                .str.replace(" ","_"))
```

```python
#now we going to check outlier detection
dataset[["price_per_unit","quantity","total_spent"]].describe()
```

```python
Q1=dataset["total_spent"].quantile(0.25)
Q3=dataset["total_spent"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
print("Lower:",lower)
print("Upper:",upper)

outlier=dataset[(dataset["total_spent"]<lower)|(dataset["total_spent"]>lower)]
```

```python
outlier = dataset[
    (dataset["total_spent"] < lower) |
    (dataset["total_spent"] > upper)
]

outlier[[
    "transaction_id",
    "price_per_unit",
    "quantity",
    "total_spent"
]]
```

```python
#Total Revenue
total_revenue=dataset["total_spent"].sum()
print("Total Revenue.",total_revenue)
```

```python
#Average Order value(AOV)
aov=dataset["total_spent"].sum()/dataset["transaction_id"].nunique()
print("Average Order value:",aov)
```

```python
#"Month with highest sales"
dataset["transaction_date"] = pd.to_datetime(dataset["transaction_date"])

monthly_sales = dataset.groupby(
    dataset["transaction_date"].dt.to_period("M")
)["total_spent"].sum()

print(monthly_sales)

print("Highest Sales Month:", monthly_sales.idxmax())
print("Highest Sales:", monthly_sales.max())
```

```python
# Category Generating the most revenue
category_revenue=dataset.groupby("category")["total_spent"].sum()
print("category_revenue")

print("Top Category:",category_revenue.idxmax())
print("Revenue:",category_revenue.max())
```

```python
#daily sales
daily_sales=dataset.groupby(dataset["transaction_date"].dt.date)["total_spent"].sum()
print(daily_sales)
```

```python
#weekly sales
weekly_sales=dataset.groupby(dataset["transaction_date"].dt.to_period("W"))["total_spent"].sum()
print(weekly_sales)
```

```python
#Monthly sales
monthly_sales=dataset.groupby(dataset["transaction_date"].dt.to_period("M"))["total_spent"].sum()
print(monthly_sales)
```

```python
#Quaterly Sales
quarterly_sales=dataset.groupby(dataset["transaction_date"].dt.to_period("Q"))["total_spent"].sum()
print(quarterly_sales)
```
